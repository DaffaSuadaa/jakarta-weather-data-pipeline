```python
import requests
import pandas as pd
from datetime import datetime
```

```python
# 1. TEST EXTRACT
url = "https://api.open-meteo.com/v1/forecast?latitude=-6.2146&longitude=106.8451&current_weather=true"
response = requests.get(url)

if response.status_code == 200:
    data = response.json()['current_weather']
    print("Koneksi API Berhasil!")
    print(data)
else:
    print("Gagal mengambil data!")
```

```python
# 2. TEST TRANSFORM
df = pd.DataFrame([data])

# Merapikan format waktu agar lebih manusiawi
df['time'] = pd.to_datetime(df['time'])
df['extracted_at'] = datetime.now()

# Menampilkan tabel
df.head()
```

```python
from sqlalchemy import create_engine

# Hubungkan ke Postgres Docker (Port 5433)
# Format: postgresql://user:password@localhost:port/dbname
engine = create_engine('postgresql://admin:12345@localhost:5433/weather_db')

# Coba kirim DataFrame (df) yang sudah kamu buat tadi ke tabel bernama 'test_weather'
try:
    df.to_sql('jakarta_weather', engine, if_exists='append', index=False)
    print("Berhasil! Data sudah masuk ke PostgreSQL di Docker.")
except Exception as e:
    print(f"Gagal konek ke Database: {e}")
```
